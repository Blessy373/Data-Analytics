```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn import linear_model
import numpy as np
```

```python
df=pd.read_csv('/content/home.csv')
```

```python
df
```

```python
df.isna().sum()
```

```python
df['bedrooms'].median()
```

```python
df['bedrooms']=df['bedrooms'].fillna(df['bedrooms'].median())
```

```python
df.isnull().sum()
```

```python
df_in=df.drop('price',axis=1)
```

```python
p=df['price']
```

```python
lr=linear_model.LinearRegression()
lr.fit(df_in,p)
```

```python
lr.coef_
```

```python
lr.intercept_
```

```python
lr.predict([[3000,3,40]])
```

```python
q=pd.DataFrame([[3000,3,40]],columns=['area','bedrooms','age'])
ans=lr.predict(q)
ans
```
