```python
from sklearn.datasets import load_iris
```

```python
flow=load_iris()
```

#visualisation

```python
import matplotlib.pyplot as plt
plt.gray()
for i in range(5):
    plt.matshow(flow.images[i])
    plt.show()
```

```python
import pandas as pd
df=pd.DataFrame(flow.data)
```

```python
df['target']=flow.target
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(flow.data,flow.target,test_size=0.2)
```

```python
from sklearn.ensemble import RandomForestClassifier
model=RandomForestClassifier()
model.fit(x_train,y_train)
```

```python
model.score(x_test,y_test)
```

```python
model.score(x_train,y_train)
```

```python
y_pred=model.predict(x_test)
```

```python
from sklearn.metrics import confusion_matrix
cm=confusion_matrix(y_test,y_pred)
cm
```

```python
import seaborn as sns
plt.figure(figsize=(10,7))
sns.heatmap(cm,annot=True,fmt='d')
plt.xticks(range(3),flow.target_names)
plt.yticks(range(3),flow.target_names)
plt.title('Confusion Matrix')
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.show()
```
