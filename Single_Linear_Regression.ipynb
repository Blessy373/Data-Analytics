```python
import pandas as pd
from sklearn import linear_model
import matplotlib.pyplot as plt
import numpy as np
```

```python
df=pd.read_csv('/content/homeprices.csv')
df
```

```python
plt.xlabel("area")
plt.ylabel('price')
plt.scatter(df.area,df.price,color="red",marker='*')
```

```python
new_df=df.drop('price',axis=1)
new_df
```

```python
df_price = df['price']
df_price
```

```python
lr = linear_model.LinearRegression()
# Train the model using 'area' as input and 'price' as the target variable
lr.fit(new_df, df_price)
```

```python
lr.predict([[4000]])
```

```python
# Predict using a DataFrame with correct feature names to avoid the warning
input_df = pd.DataFrame([[3300]], columns=['area'])
predicted_price = lr.predict(input_df)
print(f"Predicted price: {predicted_price[0]}")
```

```python
inp=pd.DataFrame([[4000]],columns=['area'])
predicted=lr.predict(inp)
predicted
```

```python
lr.coef_
```

```python
lr.intercept_
```

```python
df_area=pd.read_csv('/content/areas.csv')
```

```python
df_area.head(3)
```

```python
p=lr.predict(df_area)
p
```

```python
df_area['prices']=p
df_area['prices']
```

```python
df_area.to_csv("price.csv")
```
