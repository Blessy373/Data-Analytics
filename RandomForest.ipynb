```python
from sklearn.datasets import load_digits
```

```python
digi =load_digits()
```

Visualization

```python
import matplotlib.pyplot as plt
plt.gray()
for i in range(3):
  plt.matshow(digi.images[i])
```

```python
import pandas as pd
df=pd.DataFrame(digi.data)
df[0:5]
```

```python
df['target']=digi.target
```

Test,train split

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(digi.data,digi.target,test_size=0.2)
```

```python
from sklearn.ensemble import RandomForestClassifier
R=RandomForestClassifier()
R.fit(x_train,y_train)
```

```python
R.score(x_test,y_test)
```

```python
y_pred=R.predict(x_test)
```

```python
from sklearn.metrics import confusion_matrix
cm=confusion_matrix(y_test,y_pred)
cm
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 7))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=digi.target_names, yticklabels=digi.target_names)
plt.xlabel('Predicted')
plt.ylabel('Truth')
plt.title('Confusion Matrix Heatmap')
plt.show()
```
