Using Sklearn

```python
from numpy._core.numerictypes import int32
from sklearn.preprocessing import OneHotEncoder
o=OneHotEncoder(drop='first',sparse_output=False,dtype=int32)
```

```python
o.fit(df[['town']])
```

```python
Encoded=o.transform(df[['town']])
Encoded
```

```python
df_en=pd.DataFrame(Encoded,columns=['robinsville','west windsor'])
df_en
```

```python
df_en.reset_index(drop=True,inplace=True)
df_en
```

```python
df_en=pd.concat((df_en,df['price'],df['area']),axis=1)
df_en
```

```python
ler=LinearRegression()
ler.fit(df_en.drop('price',axis=1),df_en['price'])
```

```python
ler.predict([[0,1,3400]])
```

```python
ler.score(df_en.drop('price',axis=1),df_en['price'])
```
