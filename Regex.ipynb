```python
age=get_pattern(r'age (\d+)',text)
age
```

```python
name=get_pattern(r'Born(.*)\n',text).strip()
name
```

```python
year=get_pattern(r'Born.*\n(.*)\(',text).strip()
year
```

```python
place=get_pattern(r'age.*\n(.*)',text).strip()
place
```

```python
def extract_personal_information(text):
    age = get_pattern_match(r'age (\d+)', text)
    full_name = get_pattern_match(r'Born(.*)\n', text)
    birth_date = get_pattern_match(r'Born.*\n(.*)\(age', text)
    birth_place = get_pattern_match(r'\(age.*\n(.*)', text)
    return {
        'age': int(age),
        'name': full_name.strip(),
        'birth_date': birth_date.strip(),
        'birth_place': birth_place.strip()
    }
```

```python
extract_personal_information(text)
```
